Correlation plots.  
Pairwise comparison of BaseMean per transcript between the softwares
(=mean normalized expression estimates produced by DESeq2)

```python
from scipy.stats import pearsonr, spearmanr
#funtion to extract pearson and spearman R

def report_correlation(x, y, label):
    pearson_r, _ = pearsonr(x, y)
    spearman_r, _ = spearmanr(x, y)
    print(label)
    print(f"  Pearson r  = {pearson_r:.3f}")
    print(f"  Spearman ρ = {spearman_r:.3f}\n")
```

```python
# function to extract basemean 
def extract_basemean(df, id_col, label):
    bm = df[[id_col, "baseMean"]].copy()
    bm = bm.rename(columns={"baseMean": f"baseMean_{label}"})
    return bm
```

```python
#extract basemean from all three 
star_bm = extract_basemean(star_DE, "transcript_id", "STAR")
salmon_map_bm = extract_basemean(salmon_map_DE, "transcript_id", "Salmon_map")
salmon_align_bm = extract_basemean(salmon_align_DE, "transcript_id", "Salmon_align")

# merge on transcript_id
merged_bm = (
    star_bm
    .merge(salmon_map_bm, on="transcript_id", how="inner")
    .merge(salmon_align_bm, on="transcript_id", how="inner")
)

print("Number of shared transcripts:", merged_bm.shape[0])
merged_bm.head()
```

```python
#log10 transformation to handel RNA data

for col in ["baseMean_STAR", "baseMean_Salmon_map", "baseMean_Salmon_align"]:
    merged_bm[f"log10_{col}"] = np.log10(merged_bm[col] + 1)

merged_bm
```

STAR + featureCounts vs Salmon_map

```python
x = merged_bm["log10_baseMean_STAR"]
y = merged_bm["log10_baseMean_Salmon_map"]

pearson_r, _ = pearsonr(x, y)
spearman_r, _ = spearmanr(x, y)
r2 = pearson_r ** 2

fig = px.scatter(
    merged_bm,
    x="log10_baseMean_STAR",
    y="log10_baseMean_Salmon_map",
    opacity=0.4,
    trendline="ols",
    hover_name="transcript_id",
    labels={
        "log10_baseMean_STAR": "STAR - log10(BaseMean + 1)",
        "log10_baseMean_Salmon_map": "Salmon (map) - log10(BaseMean + 1)"
    },
    title="Overall expression agreement: STAR vs. Salmon (map)",
)

fig.add_annotation(
    x=0.05,
    y=0.95,
    xref="paper",
    yref="paper",
    showarrow=False,
    align="left",
    text=(
        f"Pearson r = {pearson_r:.3f}<br>"
        f"Spearman ρ = {spearman_r:.3f}<br>"
        f"R² = {r2:.3f}"
    )
)


fig.update_traces(
    line=dict(color="red", width=1),
    selector=dict(mode="lines")
)

fig.show()

print("STAR vs Salmon (map)")
print(f"  Pearson r  = {pearson_r:.3f}")
print(f"  Spearman ρ = {spearman_r:.3f}")
print(f"  R²         = {r2:.3f}\n")
```
